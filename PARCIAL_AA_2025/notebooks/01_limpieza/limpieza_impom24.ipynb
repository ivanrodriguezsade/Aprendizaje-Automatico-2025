# MANEJO DE DATOS FALTANTES (NaN) Y ATIPICOS
## a-Valores nulos (NaN)

### Es raro encontrar NaN en este dataset, pero si aparecen en las columnas de valor, lo más seguro es reemplazarlos por cero (0), asumiendo que un valor faltante significa que el costo no aplicó o fue cero en ese registro.

```python
# Usando los nombres de columna ACTUALES
columnas_valor_actuales = [
    'Peso_Neto_kg', 
    'FOB(USD)', 
    'Flete(USD)', 
    'Seguro(USD)', 
    'CIF(USD)'
]

# Rellenar valores nulos con 0 en las columnas de valor
df[columnas_valor_actuales] = df[columnas_valor_actuales].fillna(0)

print(f"\n✅ Valores nulos en columnas de valor imputados con 0.")
print(f"\nConteo de valores nulos después de la limpieza: \n{df.isnull().sum()}")
```

## b-Registros irrelevantes o cero

### En comercio exterior, es común tener registros donde el Valor CIF es positivo, pero el Peso Neto es cero. Esto ocurre con productos digitales (software, licencias) o bienes de poco peso/alto valor.
### Para un análisis de valor, se pueden mantener. Para modelos que dependan de la densidad o el precio por kg, hay que considerar eliminarlos o manejarlos por separado.

```python
# Filtramos registros donde todas las variables de valor sean cero, ya que no aportan a un modelo de predicción.
df_limpio = df[
    (df['CIF(USD)'] > 0) | 
    (df['FOB(USD)'] > 0)
].copy()

# Opcional: Se aislan registros de Peso Neto 0
# Si queremos hacer un análisis posterior de alto valor/cero peso
df_alto_valor_cero_peso = df_limpio[df_limpio['Peso_Neto_kg'] == 0].copy()

print(f"\nTotal de registros originales: {len(df)}")
print(f"Total de registros limpios (CIF o FOB > 0): {len(df_limpio)}")
```

# CREACION DE VARIABLES
### Crearemos las variables auxiliares más útiles para cualquier modelo predictivo en comercio exterior: el precio unitario (por kg) y el costo logístico como porcentaje.

## a-Precio unitario (USD/KG)
### (Esta es una métrica clave para el ML, ya que le enseña al modelo el "valor intrínseco" del producto.)

```python
# Evitar división por cero: si Peso_Neto_kg es 0, el Precio_Unitario será NaN.
# Si es 0, podemos asignar un valor muy alto para modelarlo como atípico, o simplemente dejarlo NaN.
# En este caso, lo dejamos NaN y lo gestionamos luego.

df_limpio['Precio_Unitario_USD_kg'] = df_limpio['FOB(USD)'] / df_limpio['Peso_Neto_kg']

# Rellenamos los NaN generados (donde Peso_Neto_kg era 0) con un valor que represente su valor único,
# como el valor FOB mismo (es decir, el costo por unidad es su costo total) o simplemente 0, 
# dependiendo de cómo se trate el subconjunto de "cero peso".

# Usaremos un gran número (como el máximo) o simplemente 0. Vamos con 0 por simplicidad, 
# asumiendo que los registros con peso 0 se analizan por separado o se excluyen de este cálculo.
df_limpio['Precio_Unitario_USD_kg'] = df_limpio['Precio_Unitario_USD_kg'].fillna(0)
```

## b-Creación de categorías NCM

### Agruparemos el NCM de 8 dígitos en categorías de 2 y 4 dígitos, lo cual es útil para análisis y modelos más amplios.

```python
# NCM de 2 dígitos (Capítulo)
df_limpio['NCM_2_digitos'] = df_limpio['NCM_8_digitos'].str[:2]

# NCM de 4 dígitos (Partida)
df_limpio['NCM_4_digitos'] = df_limpio['NCM_8_digitos'].str[:4]
```

## c-Costo logístico como porcentaje

### El objetivo de calcular este porcentaje es desacoplar el costo del flete/seguro del valor absoluto de la mercancía. Esto le enseña al modelo la "eficiencia logística" de la importación

```python
# ---Costo logístico como porcentaje ---

# 1. Definir el costo logístico (Flete + Seguro)
# Usamos los nombres de columna REALES que tienes: Flete(USD) y Seguro(USD)
costo_logistico_total = df_limpio['Flete(USD)'] + df_limpio['Seguro(USD)']

# 2. Calcular el porcentaje: (Costo Logístico / Valor FOB) * 100
# Agregamos una pequeña constante (1e-6) al divisor para evitar la división por cero,
# ya que asumimos que si FOB(USD) es 0, el Costo Logístico no es relevante para el %
# (aunque ya filtramos donde FOB(USD) era > 0).

df_limpio['Costo_Logistico_Pct'] = (
    costo_logistico_total / (df_limpio['FOB(USD)'] + 1e-6)
) * 100

# 3. Inspeccionar la nueva variable
print("✅ Variable Costo_Logistico_Pct creada.")
print("\nPrimeros 5 valores de las variables logísticas:")
print(df_limpio[['FOB(USD)', 'Flete(USD)', 'Seguro(USD)', 'Costo_Logistico_Pct']].head())
```

# UNION DE TABLAS DE REFERENCIAS (MERGE)

```python
# --- UNION DE TABLAS DE REFERENCIAS ---
print("Iniciando Merge Final sobre df_limpio...")

# Paso A: Unir Nombres de PAÍS
# df_paises_referencia debe tener 'codigo_pais' y 'nombre_pais'
df_limpio = df_limpio.merge(
    df_paises_referencia, 
    left_on='Pais_Origen_Cod',
    right_on='codigo_pais',
    how='left' # Mantener todos los registros del df_limpio
)
# Eliminamos la clave duplicada de la tabla de referencia
df_limpio.drop(columns=['codigo_pais'], inplace=True)
print("✅ Nombres de País unidos.")


# Paso B: Unir Descripciones NCM
# df_ncm_referencia debe tener 'NCM_8_digitos' y 'nombre_mercaderia'
df_limpio = df_limpio.merge(
    df_ncm_referencia,
    left_on='NCM_8_digitos',
    right_on='NCM_8_digitos',
    how='left' # Mantener todos los registros del df_limpio
)
print("✅ Nombres de Mercadería (8 dígitos) unidos.")


# --- Verificación de Resultados y Cobertura ---
print("\n" + "="*70)
print("✨ RESULTADO FINAL DE UNIÓN Y VERIFICACIÓN ✨")
print("======================================================================")

# 1. Revisamos las nuevas columnas
print("Primeras 10 filas con las columnas añadidas:")
print(df_limpio[['Pais_Origen_Cod', 'nombre_pais', 
                 'NCM_8_digitos', 'nombre_mercaderia', 
                 'Precio_Unitario_USD_kg']].head(10)) 

# 2. Verificar el porcentaje de éxito en la unión NCM
porcentaje_match_ncm = df_limpio['nombre_mercaderia'].notna().mean() * 100
print(f"\nPorcentaje de NCM que hicieron Match (Cobertura): {porcentaje_match_ncm:.2f}%")

# 3. Verificar el porcentaje de éxito en la unión de Países
porcentaje_match_pais = df_limpio['nombre_pais'].notna().mean() * 100
print(f"Porcentaje de Países que hicieron Match (Cobertura): {porcentaje_match_pais:.2f}%")

print(f"\nDimensiones finales del dataset: {df_limpio.shape}")
print("======================================================================")
```

```python
# --- PASO EXTRA: Relleno de Descripciones NCM Faltantes (Mejora de Cobertura) ---

print("\nIniciando relleno jerárquico de descripciones NCM...")

# 1. Preparar la tabla de referencia NCM para diferentes niveles

# Crear una copia de la referencia NCM para no modificar la original
df_ncm_fill = df_ncm_referencia.copy()

# Rellenar las posibles descripciones vacías/erróneas con un NaN real para poder unirlas
df_ncm_fill['nombre_mercaderia'] = df_ncm_fill['nombre_mercaderia'].replace('#¿NOMBRE?', pd.NA)


# --- 2. Preparar el DataFrame de Partidas (4 dígitos) para la unión ---
df_partida_ref = df_ncm_fill[['NCM_8_digitos', 'nombre_mercaderia']].copy()

# Creamos la clave de 4 dígitos (Partida)
df_partida_ref['NCM_4_digitos'] = df_partida_ref['NCM_8_digitos'].str[:4]

# Solo queremos la descripción más representativa por Partida
df_partida_ref.drop_duplicates(subset=['NCM_4_digitos'], keep='first', inplace=True)
df_partida_ref.rename(columns={'nombre_mercaderia': 'desc_4_digitos'}, inplace=True)


# --- 3. Unir Partida (4 dígitos) a los registros de df_limpio ---
# Creamos la clave de 4 dígitos en df_limpio
df_limpio['NCM_4_digitos'] = df_limpio['NCM_8_digitos'].str[:4]

df_limpio = df_limpio.merge(
    df_partida_ref[['NCM_4_digitos', 'desc_4_digitos']],
    on='NCM_4_digitos',
    how='left'
)

# --- 4. Rellenar los NaN de NCM de 8 dígitos con la descripción de 4 dígitos ---
# Rellenamos solo si 'nombre_mercaderia' era NaN
df_limpio['nombre_mercaderia'] = df_limpio['nombre_mercaderia'].fillna(df_limpio['desc_4_digitos'])


# --- 5. Limpieza Final y Verificación ---
df_limpio.drop(columns=['NCM_4_digitos', 'desc_4_digitos'], inplace=True) # Eliminamos las columnas auxiliares

# Re-verificación de la cobertura
porcentaje_final = df_limpio['nombre_mercaderia'].notna().mean() * 100

print(f"\n✅ Cobertura de NCM después del relleno jerárquico (Partida 4 dígitos): {porcentaje_final:.2f}%")
print("======================================================================")
```

```python
# --- PASO FINAL: Relleno de Descripciones NCM por Capítulo (2 Dígitos) ---

print("Intentando relleno final por Capítulo (2 dígitos)...")

# 1. Preparar el DataFrame de Capítulos (2 dígitos) para la unión
df_capitulo_ref = df_ncm_referencia.copy()

# Rellenar las posibles descripciones vacías/erróneas con un NaN real (por si acaso)
df_capitulo_ref['nombre_mercaderia'] = df_capitulo_ref['nombre_mercaderia'].replace('#¿NOMBRE?', pd.NA)

# Creamos la clave de 2 dígitos (Capítulo)
df_capitulo_ref['NCM_2_digitos'] = df_capitulo_ref['NCM_8_digitos'].str[:2]

# Solo queremos la descripción más representativa por Capítulo
df_capitulo_ref.drop_duplicates(subset=['NCM_2_digitos'], keep='first', inplace=True)
df_capitulo_ref.rename(columns={'nombre_mercaderia': 'desc_2_digitos'}, inplace=True)


# 2. Unir Capítulo (2 dígitos) a los registros de df_limpio
# Creamos la clave de 2 dígitos en df_limpio (si no se creó antes)
if 'NCM_2_digitos' not in df_limpio.columns:
    df_limpio['NCM_2_digitos'] = df_limpio['NCM_8_digitos'].str[:2]

df_limpio = df_limpio.merge(
    df_capitulo_ref[['NCM_2_digitos', 'desc_2_digitos']],
    on='NCM_2_digitos',
    how='left'
)

# 3. Rellenar los NaN que QUEDAN en 'nombre_mercaderia' con la descripción de capítulo
df_limpio['nombre_mercaderia'] = df_limpio['nombre_mercaderia'].fillna(df_limpio['desc_2_digitos'])


# 4. Limpieza Final y Verificación
df_limpio.drop(columns=['desc_2_digitos', 'NCM_2_digitos'], inplace=True) # Eliminamos las columnas auxiliares

# Re-verificación de la cobertura FINAL
porcentaje_final = df_limpio['nombre_mercaderia'].notna().mean() * 100

print(f"\n✅ Cobertura de NCM TOTAL FINAL: {porcentaje_final:.2f}%")
print("======================================================================")
```

```python
# --- PASO FINAL DE IMPUTACIÓN Y LIMPIEZA ---

# 1. Imputar los NaN restantes en nombre_mercaderia
df_limpio['nombre_mercaderia'] = df_limpio['nombre_mercaderia'].fillna('Sin_Descripcion_NCM')

# 2. Imputar los NaN restantes en nombre_pais (si quedan, que será menos del 0.01%)
df_limpio['nombre_pais'] = df_limpio['nombre_pais'].fillna('Pais_Desconocido')

print("\n✅ Imputación final de NaN de variables categóricas completada.")
print("El dataset está listo para el análisis y Machine Learning.")
```
